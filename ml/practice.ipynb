```python
import numpy as np
from scipy.optimize import fsolve

def solve_for_t(X, B_mat, C_mat):
    # Define the function: f(t) = (A * B * C) - X_target
    # We want to find t where f(t) = 0
    def objective(t):
        A = np.array([[t, 1 - t]])
        # Perform matrix multiplication
        result = A @ B_mat @ C_mat
        return result.item() - X

    t_solution = fsolve(objective, x0=0.0)
    
    return t_solution[0]
```

```python
# Example usage:

B = np.array([[1, 2], [3, 4]])
C = np.array([[5], [6]])
target = 50

t_sol  = solve_for_t(target, B, C)
A_final = np.array([t_sol, 1 - t_sol])
print("A_final:", A_final)
```
